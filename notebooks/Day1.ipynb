```python
from pathlib import Path

data_root = Path("/content/drive/MyDrive/leukemia_project/TXL-PBC")

print(data_root.exists())
print(len(list((data_root / "images/train").glob("*.png"))))
```

```python
yaml_text = f"""
path: {data_root}
train: images/train
val: images/val

names:
  0: WBC
  1: RBC
  2: Platelet
"""

Path("/content/txl_pbc.yaml").write_text(yaml_text)
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

model.train(
    data="/content/txl_pbc.yaml",
    epochs=20,
    imgsz=640,
    batch=16,
    device=0,
    workers=2,
    project="/content/runs",
    name="txl_pbc_baseline",
    patience=5,
)
```
